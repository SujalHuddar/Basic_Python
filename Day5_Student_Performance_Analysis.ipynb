```python
# Overall performance statistics
print("Average final percentage")
print(df['Final_Percentage'].mean())

print("Median final percentage")
print(df['Final_Percentage'].median())

print("Highest final percentage")
print(df['Final_Percentage'].max())

print("lowest final persentage")
print(df['Final_Percentage'].min())
```

```python
# Count students by performance level
print("Students by performance level")
print(df['Performance_Level'].value_counts())
```

```python
# Count students who passed and failed
print("Pass or fail destribution")
print(df['Pass_Fail'].value_counts())
```

```python
# create a bar chart of student counts by Performance_Level using Seaborn.
performance_count=df['Performance_Level'].value_counts()
print(performance_count)

sns.barplot(x=performance_count.index, y=performance_count.values)
plt.title('student counts by Performance_Level')
plt.xlabel('Performance_level')
plt.ylabel('number of students')
plt.show()
```

```python
# Investigate whether average daily study hours differ across the four Performance_Level categories.
average_study_hour = df.groupby('Performance_Level')['Study_Hours_Per_Day'].mean()
print(average_study_hour)

sns.barplot(x=average_study_hour.index, y=average_study_hour.values)
plt.title('average daily study hours by Performance_Level')
plt.xlabel('Performance level')
plt.ylabel('average study hours per day')
plt.show()
```

```python
# Calculate the average attendance percentage for each Performance_Level.
avg_attendence_percentage = df.groupby('Performance_Level')['Attendance_Percentage'].mean()
avg_attendence_percentage
```

```python
# Create a Seaborn bar chart using average_attendance.index and average_attendance.values, with a suitable title and axis labels.
sns.barplot(x = avg_attendence_percentage.index, y=avg_attendence_percentage.values)
plt.title('Average attendance percentage for each Performance_Level')
plt.xlabel('Performance level')
plt.ylabel('Average attendence')
plt.show()
```

```python
# Calculate average score for each subject
average_score = df[['Math_Score','Science_Score','English_Score']].mean()
average_score
```

```python
# create a bar chart comparing the average Math, Science, and English scores.
plt.bar(average_score.index, average_score.values)
plt.title('Average math, science, and english scores')
plt.xlabel('subjects')
plt.ylabel('Average scores')
plt.show()
```

```python
# investigate whether daily study hours are associated with students' final percentages.
correlation = df['Study_Hours_Per_Day'].corr(df['Final_Percentage'])
correlation
```

```python
# the relationship between study hours and final percentage.
sns.scatterplot(x='Study_Hours_Per_Day', y ='Final_Percentage', data = df)
plt.title('study hours vs final percentage')
plt.xlabel('study hours')
plt.ylabel('final persentage')
plt.show()
```

```python
# Student Performance Analysis — Key Findings

print("STUDENT PERFORMANCE ANALYSIS")
print("-" * 35)

print("Total students:", len(df))
print("Average final percentage:", round(df['Final_Percentage'].mean(), 2))
print("Pass rate (%):", round((df['Pass_Fail'] == 'Pass').mean() * 100, 2))

print("\nStudents by performance level:")
print(df['Performance_Level'].value_counts())

print("\nAverage subject scores:")
print(df[['Math_Score', 'Science_Score', 'English_Score']].mean().round(2))

print("\nStudy hours vs final percentage correlation:")
print(round(df['Study_Hours_Per_Day'].corr(df['Final_Percentage']), 3))
```
