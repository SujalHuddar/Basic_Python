```python
# Overall performance statistics
print("Average final percentage")
print(df['Final_Percentage'].mean())

print("Median final percentage")
print(df['Final_Percentage'].median())

print("Highest final percentage")
print(df['Final_Percentage'].max())

print("lowest final persentage")
print(df['Final_Percentage'].min())
```

```python
# Count students by performance level
print("Students by performance level")
print(df['Performance_Level'].value_counts())
```

```python
# Count students who passed and failed
print("Pass or fail destribution")
print(df['Pass_Fail'].value_counts())
```

```python
# create a bar chart of student counts by Performance_Level using Seaborn.
performance_count=df['Performance_Level'].value_counts()
print(performance_count)

sns.barplot(x=performance_count.index, y=performance_count.values)
plt.title('student counts by Performance_Level')
plt.xlabel('Performance_level')
plt.ylabel('number of students')
plt.show()
```

```python
# Investigate whether average daily study hours differ across the four Performance_Level categories.
average_study_hour = df.groupby('Performance_Level')['Study_Hours_Per_Day'].mean()
print(average_study_hour)

sns.barplot(x=average_study_hour.index, y=average_study_hour.values)
plt.title('average daily study hours by Performance_Level')
plt.xlabel('Performance level')
plt.ylabel('average study hours per day')
plt.show()
```
