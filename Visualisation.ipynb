```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Name": ["Amit", "Rahul", "Sneha", "Priya", "Kiran", "Rohit", "Anjali", "Vikas"],
    "Department": ["CSE", "ECE", "CSE", "ISE", "ECE", "CSE", "ISE", "ECE"],
    "Marks": [78, 65, 92, 55, 88, 72, 95, 61],
    "City": ["Belgaum", "Hubli", "Belgaum", "Bangalore",
             "Hubli", "Belgaum", "Bangalore", "Hubli"]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# calculate the average marks for each department and create a bar chart showing the average marks of each department.

avg_marks = df.groupby('Department')['Marks'].mean()
avg_marks

plt.bar(avg_marks.index, avg_marks.values)
plt.title('Average marks of each department')
plt.show()
```

```python
# make a bar chart showing the number of students in each city.

city_count = df['City'].value_counts()
city_count

plt.bar(city_count.index, city_count.values)
plt.title(' Number of students in each city')
plt.show()
```

```python
# Create a box plot of Marks by Department.

sns.boxplot(x='Marks', y='Department', data = df)
plt.title('Marks by Department')
plt.show()
```

```python
# average marks for each department.

sns.barplot(x='Department', y='Marks', data=df)
plt.show()
```

```python
sns.countplot(x='City', data = df)
plt.title('Number of student in each city')
plt.show()
```

```python
#Create a bar chart showing the average marks by city.

avg_city_marks = df.groupby('City')['Marks'].mean()
avg_city_marks
```

```python
plt.bar(avg_city_marks.index, avg_city_marks.values)
plt.title('Average marks by city')
plt.show()
```
