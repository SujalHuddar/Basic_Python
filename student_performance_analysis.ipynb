```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Student_Performance_Dataset.csv")
```

```python
df
```

```python
'''Find:
1. Total number of students
2. Average marks
3. Highest marks
4. Lowest marks'''

# 1
total_students = len(df)
print("Total Number of students are",total_students)

#2
avg_per = df['Final_Percentage'].mean()
print("Average Percentage of students:",avg_per)

#3 
highest_per = df['Final_Percentage'].max()
print("Highest Percentage of students:",highest_per)

#4
lowest_per = df['Final_Percentage'].min()
print("Lowest Percentage of students:",lowest_per)
```

```python
# Find the average Final Percentage for each Performance Level.

avg_final_per = df.groupby('Performance_Level')['Final_Percentage'].mean()
avg_final_per
```
