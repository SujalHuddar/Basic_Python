```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Student_Performance_Dataset.csv")
```

```python
df
```

```python
'''Find:
1. Total number of students
2. Average marks
3. Highest marks
4. Lowest marks'''

# 1
total_students = len(df)
print("Total Number of students are",total_students)

#2
avg_per = df['Final_Percentage'].mean()
print("Average Percentage of students:",avg_per)

#3 
highest_per = df['Final_Percentage'].max()
print("Highest Percentage of students:",highest_per)

#4
lowest_per = df['Final_Percentage'].min()
print("Lowest Percentage of students:",lowest_per)
```

```python
# Find the average Final Percentage for each Performance Level.

avg_final_per = df.groupby('Performance_Level')['Final_Percentage'].mean()
avg_final_per
```

```python
# Find which Performance_Level has the highest average Final Percentage.
higest_performance_level = avg_final_per.idxmax()
higest_performance_level
```

```python
# Find how many students are in each Performance_Level.
students = df['Performance_Level'].value_counts()
students
```

```python
# Find the number of students who Passed and Failed.
pass_fail = df['Pass_Fail'].value_counts()
pass_fail
```

```python
# Find the average Final_Percentage for each Gender.
average_final_percentage = df.groupby('Gender')['Final_Percentage'].mean()
average_final_percentage
```

```python
# find the average Attendance_Percentage for each Class.
average_attendance_percentage = df.groupby('Class')['Attendance_Percentage'].mean()
average_attendance_percentage
```

```python
overall_avg_final_percentage = df['Final_Percentage'].mean()
overall_avg_final_percentage
```

```python
# Find all students whose Final_Percentage is greater than the overall average Final_Percentage.
greater_average_final_percentage = df[df['Final_Percentage']>overall_avg_final_percentage]
greater_average_final_percentage
```

```python
# find the 5 students with the highest Final_Percentage in your actual dataset.
top_5_students = df.sort_values(by = 'Final_Percentage',ascending = False).head(5)
top_5_students
```

```python
# Create a bar chart showing the average Final_Percentage for each Class.
averag_class_final_percentage = df.groupby('Class')['Final_Percentage'].mean()
averag_class_final_percentage
```

```python
plt.bar(averag_class_final_percentage.index, averag_class_final_percentage.values)
plt.title('average Final_Percentage for each Class')
plt.xlabel('Class')
plt.ylabel('average Final percentage')
plt.show()
```

```python
# reate a bar chart showing how many students passed and failed using your actual Pass_Fail column.
count_pass_fail = df['Pass_Fail'].value_counts()
print(count_pass_fail)
plt.bar(count_pass_fail.index, count_pass_fail.values)
plt.title('students passed and failed')
plt.xlabel('Result')
plt.ylabel('Number of students')
plt.show()
```

```python
# Create a bar chart showing the average Attendance_Percentage for each Class
average_attendance_percentage_class = df.groupby('Class')['Attendance_Percentage'].mean()
print(average_attendance_percentage_class)
plt.bar(average_attendance_percentage_class.index, average_attendance_percentage_class.values)
plt.title('Average Attendance_Percentage for each Class')
plt.xlabel('class')
plt.ylabel('Average Attendance Percentage')
plt.show()
```

```python
# create a histogram to see how students' Final_Percentage values are distributed.
plt.hist(df['Final_Percentage'], bins = 10, edgecolor = 'black')
plt.title('Destribution of final percentage')
plt.xlabel('Final percentage')
plt.ylabel('students')
plt.show()
```

```python
# Find the average Study_Hours_Per_Day for each Performance_Level, then create a bar chart to compare them.
average_study_hours = df.groupby('Performance_Level')['Study_Hours_Per_Day'].mean()
print(average_study_hours)
plt.bar(average_study_hours.index, average_study_hours.values)
plt.title('Average students study hours per day')
plt.xlabel('Performance level')
plt.ylabel('average study hours')
plt.show()
```
